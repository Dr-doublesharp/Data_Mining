```python
import kagglehub
import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder
```

Step 1 - Downloading Data Set

```python
path = kagglehub.dataset_download("gregorut/videogamesales")

print("Path to dataset files:", path)
```

importing dataset

```python
file_path = 'vgsales.csv'
```

display first five rows

```python
df = pd.read_csv(file_path)
print(f"--- dataset imported successfully {df.shape} ---")
display(df.head())
```

Task 1 - checking and filling missing values

```python
num_cols = df.select_dtypes(include=[np.number]).columns #taking only numbers values
print("the missing values before processing for each column are:")
print(df[num_cols].isnull().sum())

#filling missing columns
for col in num_cols:
    if df[col].isnull().sum() > 0:
        median_val = df[col].median()
        df[col] = df[col].fillna(median_val)
        print(f"Filled column '{col}' with median: {median_val}")

print("\n the missing values after processing for each column are:")
print(df[num_cols].isnull().sum())
```

Task 2 - removing duplicated rows

```python
duplicates_count = df.duplicated().sum() # gets counts of duplicated rows
print(f"number of dublicated rows before deleting are: {duplicates_count}")

df = df.drop_duplicates().reset_index(drop=True) # removing duplicated rows
print(f"number of dublicated rows after deleting are: {df.duplicated().sum()}")
```

Task 3 - Applying Label Encoding to a single categorical column

```python
target_cat_col = 'Genre' # picking game Genre column

le = LabelEncoder()

df[target_cat_col] = df[target_cat_col].fillna('Unknown') # filling missing values, because it makes errors

df[target_cat_col + '_Encoded'] = le.fit_transform(df[target_cat_col]) # applying transform + making a new column for it

print(f"Label Encoding successfully applied to column: '{target_cat_col}'")

# showing data before and after lable encoding
display(df[[target_cat_col, target_cat_col + '_Encoded']].head(10))
```
